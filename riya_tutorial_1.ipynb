```python
#medium question

# Install once if needed
# !pip install qiskit qiskit-aer matplotlib

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# Create a 1-qubit circuit with 1 classical bit
qc = QuantumCircuit(1, 1)

# Apply Hadamard Gate
qc.h(0)

# Measure the qubit
qc.measure(0, 0)

print("Quantum Circuit:")
print(qc)

# Run the circuit on the simulator
simulator = AerSimulator()
job = simulator.run(qc, shots=1024)

# Get the result
result = job.result()
counts = result.get_counts()

print("\nMeasurement Counts:")
print(counts)

# Plot histogram
plot_histogram(counts)
plt.show()
```

```python
#hard question

# Install once if needed
# !pip install qiskit qiskit-aer matplotlib

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# Create a 3-qubit circuit with 3 classical bits
qc = QuantumCircuit(3, 3)

# Apply Hadamard gate to all three qubits
qc.h([0, 1, 2])

# Measure all qubits
qc.measure([0, 1, 2], [0, 1, 2])

print("Quantum Circuit:")
print(qc)

# Run the circuit on the simulator
simulator = AerSimulator()
job = simulator.run(qc, shots=1024)

# Get the results
result = job.result()
counts = result.get_counts()

print("\nMeasurement Counts:")
print(counts)

# Plot histogram
plot_histogram(counts)
plt.show()

# Print theoretical probability
print("\nTheoretical Probability for each state = 1/8 = 0.125 (12.5%)")
```

```python
# Install once if needed
# !pip install qiskit qiskit-aer matplotlib

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
import matplotlib.pyplot as plt
import random

# Number of shots
shots = 1024

# ---------------- Quantum Random Number Generator ----------------

# Create a 1-qubit circuit
qc = QuantumCircuit(1, 1)

# Apply Hadamard gate to create superposition
qc.h(0)

# Measure the qubit
qc.measure(0, 0)

# Run on simulator
simulator = AerSimulator()
job = simulator.run(qc, shots=shots)
result = job.result()

# Get counts
quantum_counts = result.get_counts()

print("Quantum Random Number Generator Results:")
print(quantum_counts)

# Plot Quantum Histogram
plt.figure(figsize=(6,4))
plt.bar(quantum_counts.keys(), quantum_counts.values())
plt.title("Quantum Random Number Generator")
plt.xlabel("Bit")
plt.ylabel("Counts")
plt.grid(axis='y')
plt.show()


# ---------------- Python Pseudo-Random Number Generator ----------------

python_counts = {'0': 0, '1': 0}

for i in range(shots):
    bit = random.randint(0, 1)
    python_counts[str(bit)] += 1

print("\nPython Pseudo-Random Generator Results:")
print(python_counts)

# Plot Python Histogram
plt.figure(figsize=(6,4))
plt.bar(python_counts.keys(), python_counts.values())
plt.title("Python Pseudo-Random Generator")
plt.xlabel("Bit")
plt.ylabel("Counts")
plt.grid(axis='y')
plt.show()


# ---------------- Comparison ----------------

print("\nComparison")
print("----------------------------")
print("Quantum Generator :", quantum_counts)
print("Python Generator  :", python_counts)
```
