```python
#easy
!pip install qiskit qiskit-aer

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

# Create a 2-qubit circuit
qc = QuantumCircuit(2, 2)

# Construct Phi+ Bell state
qc.h(0)
qc.cx(0, 1)

# Measure both qubits
qc.measure([0, 1], [0, 1])

# Simulate for 1024 shots
simulator = AerSimulator()
result = simulator.run(qc, shots=1024).result()
counts = result.get_counts()

print("Measurement results:", counts)

# Verify only 00 and 11 occur
if set(counts.keys()).issubset({'00', '11'}):
    print("Verified: Only 00 and 11 outcomes occur.")
else:
    print("Unexpected measurement outcomes.")
```

```python
#medium
!pip install qiskit qiskit-aer

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

simulator = AerSimulator()

def create_bell_state(state):
    qc = QuantumCircuit(2, 2)

    # Create the four Bell states
    if state == "Phi+":
        qc.h(0)
        qc.cx(0, 1)

    elif state == "Phi-":
        qc.x(0)
        qc.h(0)
        qc.cx(0, 1)
        qc.z(0)

    elif state == "Psi+":
        qc.h(0)
        qc.cx(0, 1)
        qc.x(1)

    elif state == "Psi-":
        qc.h(0)
        qc.cx(0, 1)
        qc.x(1)
        qc.z(0)

    qc.measure([0, 1], [0, 1])
    return qc

bell_states = ["Phi+", "Phi-", "Psi+", "Psi-"]

print("Bell State Measurement Distributions\n")

for state in bell_states:
    qc = create_bell_state(state)
    result = simulator.run(qc, shots=1024).result()
    counts = result.get_counts()

    print(state, ":", counts)
```

```python
#hard
!pip install qiskit qiskit-aer

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

# Create Bell state Phi+
qc = QuantumCircuit(2, 2)

qc.h(0)
qc.cx(0, 1)

# Convert Hadamard basis measurement to computational basis
qc.h(0)
qc.h(1)

# Measure both qubits
qc.measure([0, 1], [0, 1])

# Simulate 1024 shots
simulator = AerSimulator()
result = simulator.run(qc, shots=1024).result()
counts = result.get_counts()

print("Hadamard basis measurement results:", counts)
```

```python
#challenge
!pip install qiskit qiskit-aer

import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, partial_trace, entropy

# ---------- GHZ STATE ----------
ghz = QuantumCircuit(3)
ghz.h(0)
ghz.cx(0, 1)
ghz.cx(0, 2)

ghz_state = Statevector.from_instruction(ghz)

print("GHZ State:")
print(ghz_state)

# Trace out qubit 0
ghz_reduced = partial_trace(ghz_state, [0])
print("\nGHZ reduced-state entropy:",
      round(entropy(ghz_reduced, base=2), 4))

# ---------- W STATE ----------
w = QuantumCircuit(3)

# Prepare W state using controlled rotations
theta = 2 * np.arccos(1 / np.sqrt(3))
w.ry(theta, 0)
w.ch(0, 1)
w.cx(1, 2)
w.x(0)
w.cx(0, 1)
w.x(0)

w_state = Statevector.from_instruction(w)

print("\nW State:")
print(w_state)

# Trace out qubit 0
w_reduced = partial_trace(w_state, [0])
print("\nW reduced-state entropy:",
      round(entropy(w_reduced, base=2), 4))
```
