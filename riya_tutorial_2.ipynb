```python
#real world tutorial 2





#Install once if needed
# !pip install qiskit qiskit-aer qiskit[visualization] matplotlib

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector
import matplotlib.pyplot as plt
import numpy as np

# Create a quantum circuit with one qubit
qc = QuantumCircuit(1)

# Put the qubit into superposition
qc.h(0)

# Add a small random phase error
phase_error = np.random.uniform(0, np.pi/8)
qc.p(phase_error, 0)

# Get the final statevector
state = Statevector.from_instruction(qc)

print("Quantum Circuit:")
print(qc)

print("\nRandom Phase Error (radians):", round(phase_error, 4))

print("\nFinal Statevector:")
print(state)

# Display Bloch Sphere
plot_bloch_multivector(state)
plt.show()
```
