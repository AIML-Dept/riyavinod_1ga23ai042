```python
# Install Qiskit (run once if needed)
!pip install qiskit qiskit-aer

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

# Create a Hadamard-superposition circuit
qc = QuantumCircuit(1, 1)

# Put the qubit into equal superposition
qc.h(0)

# Measure the qubit
qc.measure(0, 0)

# Create simulator
simulator = AerSimulator()

# Different numbers of shots
shots_list = [100, 1000, 10000]

print("Hadamard Superposition Measurement")
print("=" * 55)
print(f"{'Shots':<10}{'0 Count':<12}{'1 Count':<12}{'P(0)':<12}{'P(1)':<12}")
print("-" * 55)

for shots in shots_list:

    result = simulator.run(qc, shots=shots).result()
    counts = result.get_counts()

    count_0 = counts.get('0', 0)
    count_1 = counts.get('1', 0)

    probability_0 = count_0 / shots
    probability_1 = count_1 / shots

    print(
        f"{shots:<10}"
        f"{count_0:<12}"
        f"{count_1:<12}"
        f"{probability_0:<12.4f}"
        f"{probability_1:<12.4f}"
    )
```

```python
#medium question
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

simulator = AerSimulator()
shots = 1024

# -------------------------------
# Prepare |+> state
# -------------------------------
plus = QuantumCircuit(1, 1)

# |+> = H|0>
plus.h(0)

# Apply H to measure in X-basis
plus.h(0)
plus.measure(0, 0)

result_plus = simulator.run(plus, shots=shots).result()
counts_plus = result_plus.get_counts()

# -------------------------------
# Prepare |-> state
# -------------------------------
minus = QuantumCircuit(1, 1)

# |-> = H|1>
minus.x(0)
minus.h(0)

# Apply H to measure in X-basis
minus.h(0)
minus.measure(0, 0)

result_minus = simulator.run(minus, shots=shots).result()
counts_minus = result_minus.get_counts()

# -------------------------------
# Display results
# -------------------------------
print("X-Basis Measurement Results")
print("=" * 40)

print("\n|+> State:")
print(plus)
print("Measurement counts:", counts_plus)

print("\n|-> State:")
print(minus)
print("Measurement counts:", counts_minus)

# Probabilities
plus_0 = counts_plus.get('0', 0) / shots
plus_1 = counts_plus.get('1', 0) / shots

minus_0 = counts_minus.get('0', 0) / shots
minus_1 = counts_minus.get('1', 0) / shots

print("\nProbabilities:")
print("|+> : P(0) =", round(plus_0, 3), ", P(1) =", round(plus_1, 3))
print("|-> : P(0) =", round(minus_0, 3), ", P(1) =", round(minus_1, 3))
```

```python
#hard ques
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

simulator = AerSimulator()
shots = 1024

# Create a 2-qubit Bell pair
qc = QuantumCircuit(2, 1)

# Create Bell state (|00> + |11>) / sqrt(2)
qc.h(0)
qc.cx(0, 1)

# Partially measure only qubit 0
qc.measure(0, 0)

# Run the circuit
result = simulator.run(qc, shots=shots).result()
counts = result.get_counts()

# Display circuit and results
print("Bell Pair with Partial Measurement:")
print(qc)

print("\nMeasurement Results:")
print(counts)

# Analyse the unmeasured qubit
print("\nAnalysis:")
print("If qubit 0 is measured as 0, qubit 1 collapses to |0>.")
print("If qubit 0 is measured as 1, qubit 1 collapses to |1>.")
print("Therefore, the unmeasured qubit is perfectly correlated with the measured qubit.")
```

```python
#real world
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
import numpy as np

simulator = AerSimulator()

# Unknown rotation angle (in practice, this would be unknown)
true_angle = np.pi / 3

# Number of measurements
shots = 4096

# Create a circuit with an unknown Ry rotation
qc = QuantumCircuit(1, 1)

qc.ry(true_angle, 0)

# Measure the qubit
qc.measure(0, 0)

# Run repeated measurements
result = simulator.run(qc, shots=shots).result()
counts = result.get_counts()

# Calculate probability of measuring |1>
p1 = counts.get('1', 0) / shots

# For Ry(theta), P(1) = sin²(theta/2)
estimated_angle = 2 * np.arcsin(np.sqrt(p1))

print("Quantum Rotation Angle Estimation")
print("=" * 40)

print("Measurement counts:", counts)
print("Estimated P(1):", round(p1, 4))

print("\nActual angle:", round(true_angle, 4), "radians")
print("Estimated angle:", round(estimated_angle, 4), "radians")

print("\nEstimated angle in degrees:",
      round(np.degrees(estimated_angle), 2), "degrees")
```

```python
# challenge
import numpy as np
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from scipy.stats import binomtest

# ---------------------------------------
# Settings
# ---------------------------------------
shots = 10000
simulator = AerSimulator()

# ---------------------------------------
# Quantum Random Bit Generator
# ---------------------------------------
qc = QuantumCircuit(1, 1)
qc.h(0)
qc.measure(0, 0)

result = simulator.run(qc, shots=shots).result()
quantum_counts = result.get_counts()

quantum_ones = quantum_counts.get('1', 0)

# ---------------------------------------
# Biased Classical Pseudo-Random Generator
# ---------------------------------------
np.random.seed(42)

# Biased generator: probability of 1 = 0.60
classical_bits = np.random.choice(
    [0, 1],
    size=shots,
    p=[0.40, 0.60]
)

classical_ones = np.sum(classical_bits)

# ---------------------------------------
# Statistical Hypothesis Test
# H0: P(1) = 0.5
# H1: P(1) != 0.5
# ---------------------------------------

quantum_test = binomtest(
    quantum_ones,
    shots,
    p=0.5
)

classical_test = binomtest(
    classical_ones,
    shots,
    p=0.5
)

# ---------------------------------------
# Display Results
# ---------------------------------------
print("QUANTUM vs CLASSICAL RANDOMNESS")
print("=" * 45)

print("\nQuantum Generator")
print("Counts:", quantum_counts)
print("P(1):", round(quantum_ones / shots, 4))
print("p-value:", quantum_test.pvalue)

print("\nClassical Biased Generator")
print("Number of 1s:", classical_ones)
print("P(1):", round(classical_ones / shots, 4))
print("p-value:", classical_test.pvalue)

# ---------------------------------------
# Interpretation
# ---------------------------------------
alpha = 0.05

print("\nStatistical Decision")
print("=" * 45)

if quantum_test.pvalue > alpha:
    print("Quantum: Fail to reject H0")
    print("The data is consistent with a fair 50-50 generator.")
else:
    print("Quantum: Reject H0")
    print("The quantum generator appears biased.")

if classical_test.pvalue < alpha:
    print("Classical: Reject H0")
    print("The classical generator is statistically biased.")
else:
    print("Classical: Fail to reject H0")
    print("No significant bias detected.")
```
