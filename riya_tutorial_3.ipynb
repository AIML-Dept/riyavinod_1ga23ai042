```python
#easy

# Install Qiskit (Run only once)
# !pip install qiskit qiskit-aer

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# ---------------- X Gate ----------------
qc_x = QuantumCircuit(1)
qc_x.x(0)

state_x = Statevector.from_instruction(qc_x)

print("===== X Gate =====")
print(qc_x)
print("Statevector:")
print(state_x)
print()

# ---------------- Y Gate ----------------
qc_y = QuantumCircuit(1)
qc_y.y(0)

state_y = Statevector.from_instruction(qc_y)

print("===== Y Gate =====")
print(qc_y)
print("Statevector:")
print(state_y)
print()

# ---------------- Z Gate ----------------
qc_z = QuantumCircuit(1)
qc_z.z(0)

state_z = Statevector.from_instruction(qc_z)

print("===== Z Gate =====")
print(qc_z)
print("Statevector:")
print(state_z)
```

```python
# real world 

# Install Qiskit (Run only once)
# !pip install qiskit qiskit-aer

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
import numpy as np

# Create a single-qubit quantum circuit
qc = QuantumCircuit(1)

# Apply Hadamard gate to simulate a fair quantum coin flip
qc.h(0)

# Get the statevector
state = Statevector.from_instruction(qc)

print("Quantum Coin Flip Circuit:")
print(qc)

print("\nStatevector:")
print(state)

# Calculate probabilities
probabilities = state.probabilities()

print("\nProbability of Heads (|0>): {:.2f}".format(probabilities[0]))
print("Probability of Tails (|1>): {:.2f}".format(probabilities[1]))

# Determine the winner
if probabilities[0] > probabilities[1]:
    print("\nResult: Heads has a higher chance of winning.")
elif probabilities[1] > probabilities[0]:
    print("\nResult: Tails has a higher chance of winning.")
else:
    print("\nResult: Fair Coin! Heads and Tails each have a 50% chance of winning.")
```
